```python
import requests
from bs4 import BeautifulSoup
import re
import pandas as pd

# --- Request ---
url = "https://www.suzuki.co.id/dealers"
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}
response = requests.get(url, headers=headers, timeout=15)
response.raise_for_status()
print(f"Status: {response.status_code}")

soup = BeautifulSoup(response.text, 'html.parser')

# --- Cari semua blok dealer (div dengan class "dealer-item") ---
dealer_blocks = soup.find_all('div', class_='dealer-item')

print(f"Jumlah dealer terdeteksi: {len(dealer_blocks)}")

result = []

for block in dealer_blocks:
    try:
        # Nama Dealer
        name_tag = block.find('h4', class_='dealer-name')
        dealer = name_tag.get_text(strip=True) if name_tag else "N/A"

        # Kota
        city_tag = block.find('p', class_='dealer-city')
        city = city_tag.get_text(strip=True) if city_tag else "N/A"

        # Alamat
        alamat_tag = block.find('p', class_='dealer-address')
        alamat = alamat_tag.get_text(strip=True) if alamat_tag else "Tidak tersedia"

        # Nomor Telepon
        phone_tag = block.find('p', class_='dealer-phone')
        phone = phone_tag.get_text(strip=True) if phone_tag else "Tidak tersedia"

        # Jenis Dealer
        type_tag = block.find('p', class_='dealer-type')
        dealer_type = type_tag.get_text(strip=True) if type_tag else "Tidak tersedia"

        # Koordinat dari data attributes
        lat = block.get('data-lat', 'N/A')
        long = block.get('data-lng', 'N/A')

        # Simpan
        result.append({
            'Dealer': dealer,
            'Kota': city,
            'Alamat': alamat,
            'Nomor Telepon': phone,
            'Jenis Dealer': dealer_type,
            'Latitude': lat,
            'Longitude': long
        })

    except Exception as e:
        print(f"Error parsing dealer block: {e}")
        continue

# --- Output ---
print("\n" + "="*100)
print("CONTOH 5 DEALER PERTAMA".center(100))
print("="*100)
for item in result[:5]:
    print(f"Dealer       : {item['Dealer']}")
    print(f"Kota         : {item['Kota']}")
    print(f"Alamat       : {item['Alamat']}")
    print(f"Nomor Telepon: {item['Nomor Telepon']}")
    print(f"Jenis Dealer : {item['Jenis Dealer']}")
    print(f"Lat/Long     : {item['Latitude']}, {item['Longitude']}")
    print("-" * 100)
```

```python
# --- Simpan ke Excel ---
import pandas as pd

df = pd.DataFrame(result)

# Add headers
df.columns = ['Provinsi', 'Dealer', 'Alamat', 'Latitude', 'Longitude']

# Save to Excel
df.to_excel('honda_dealers_indonesia.xlsx', index=False)

print(f"Total {len(result)} dealer disimpan ke 'honda_dealers_indonesia.xlsx'")
```
